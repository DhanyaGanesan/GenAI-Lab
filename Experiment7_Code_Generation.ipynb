```python
from transformers import pipeline
```

```python
generator = pipeline(
    "text-generation",
    model="Salesforce/codegen-350M-mono"
)

print("Model loaded successfully!")
```

```python
prompt = """# Write a Python function to check whether a number is prime

def is_prime(n):
"""

result = generator(
    prompt,
    max_new_tokens=80,
    do_sample=False
)

print(result[0]["generated_text"])
```
