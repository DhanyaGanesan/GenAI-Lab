```python
# ============================================
# EXPERIMENT 10 - IMDB DATASET PREPARATION
# ============================================

from datasets import load_dataset
from transformers import AutoTokenizer

# Load IMDB dataset
dataset = load_dataset("stanfordnlp/imdb")

# Use the sample sizes specified in the lab manual
small_train = dataset["train"].shuffle(seed=42).select(range(2000))
small_test = dataset["test"].shuffle(seed=42).select(range(500))

print("Training samples:", len(small_train))
print("Test samples:", len(small_test))

# Load DistilBERT tokenizer
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")

# Tokenization
def tokenize(batch):
    return tokenizer(
        batch["text"],
        padding="max_length",
        truncation=True,
        max_length=128
    )

train_ds = small_train.map(tokenize, batched=True)
test_ds = small_test.map(tokenize, batched=True)

# Remove original text column
train_ds = train_ds.remove_columns(["text"])
test_ds = test_ds.remove_columns(["text"])

# Prepare datasets for PyTorch
train_ds.set_format("torch")
test_ds.set_format("torch")

print("\nDataset preparation completed successfully!")
print("Train samples:", len(train_ds))
print("Test samples:", len(test_ds))
```
