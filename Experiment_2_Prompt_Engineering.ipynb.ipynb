```python
from transformers import pipeline, set_seed
```

```python
generator = pipeline(
    "text-generation",
    model="gpt2"
)

set_seed(42)
```

```python
zero_shot_prompt = """
Classify the sentiment of this review as Positive or Negative.

Review:
The product quality is excellent!

Sentiment:
"""

result = generator(
    zero_shot_prompt,
    max_new_tokens=20,
    do_sample=False
)

print("===== Zero-Shot =====")
print(result[0]["generated_text"])
```

```python
few_shot_prompt = """
Review: I loved this movie.
Sentiment: Positive

Review: The service was slow and disappointing.
Sentiment: Negative

Review: The product quality is excellent!
Sentiment:
"""

result = generator(
    few_shot_prompt,
    max_new_tokens=20,
    do_sample=False
)

print("===== Few-Shot =====")
print(result[0]["generated_text"])
```

```python
cot_prompt = """
Q: A shop had 15 apples.
It sold 6 apples.
Then it received 10 more apples.

Let's think step by step.
"""

result = generator(
    cot_prompt,
    max_new_tokens=50,
    do_sample=False
)

print("===== Chain-of-Thought =====")
print(result[0]["generated_text"])
```
