```python
from transformers import BlipProcessor, BlipForConditionalGeneration
from PIL import Image
import requests
import torch

print("Libraries imported successfully!")
```

```python
processor = BlipProcessor.from_pretrained(
    "Salesforce/blip-image-captioning-base"
)

model = BlipForConditionalGeneration.from_pretrained(
    "Salesforce/blip-image-captioning-base"
)

print("BLIP model loaded successfully!")
```

```python
image_url = "https://images.unsplash.com/photo-1519125323398-675f0ddb6308"

raw_image = Image.open(
    requests.get(image_url, stream=True).raw
).convert("RGB")

print("Image loaded successfully!")
print("Image size:", raw_image.size)

display(raw_image)
```

```python
inputs = processor(raw_image, return_tensors="pt")

caption_ids = model.generate(
    **inputs,
    max_new_tokens=30
)

caption = processor.decode(
    caption_ids[0],
    skip_special_tokens=True
)

print("Generated Caption:", caption)
```

```python
from transformers import BlipForQuestionAnswering

vqa_model = BlipForQuestionAnswering.from_pretrained(
    "Salesforce/blip-vqa-base"
)

print("BLIP VQA model loaded successfully!")
```

```python
question = "What animal is in the picture?"

vqa_inputs = processor(
    raw_image,
    question,
    return_tensors="pt"
)

answer_ids = vqa_model.generate(
    **vqa_inputs
)

answer = processor.decode(
    answer_ids[0],
    skip_special_tokens=True
)

print("Question:", question)
print("Answer:", answer)
```
